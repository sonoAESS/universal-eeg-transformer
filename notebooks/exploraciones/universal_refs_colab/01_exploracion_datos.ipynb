## 1b - Calidad de los datos (diagnostico, sin filtraje)

Comprobacion de integridad y contenido fisiologico del dato real ya
construido (bandpass 1-45 Hz en la construccion). No se aplica ningun
filtraje: solo se diagnostica si el dato es bueno y no requiere
procesamiento adicional.

```python
# ---- 1b - Calidad de los datos (diagnostico, sin filtraje) ----------
import numpy as np
from scipy.signal import welch

sfreq = 160.0
mc = data["canonical"] if "canonical" in data.order else data[data.order[0]]
Xu = ds.refs["unipolar"]                 # (N, C) senal unipolar (ancla)
ch = ds.ch_names
N, C = Xu.shape

# 1) integridad
n_nan = int(np.isnan(Xu).sum()); n_inf = int(np.isinf(Xu).sum())
print(f"Integridad: NaN={n_nan}, Inf={n_inf}, muestras={N}, canales={C}")

# 2) canales planos / atascados
std_ch = Xu.std(axis=0)
flat = [ch[i] for i in np.where(std_ch < 1e-9)[0]]
print(f"Canales planos (std~0): {flat if flat else 'ninguno'}")

# 3) residual de artefactos post-build (|z|>6 en algun canal)
z = (Xu - Xu.mean(0)) / (Xu.std(0) + 1e-12)
frac_bad = float((np.abs(z).max(axis=1) > 6).mean())
print(f"Fraccion de muestras con |z|>6 en algun canal: {frac_bad:.4%}")

# 4) PSD media, linea 60 Hz y bandas oscilatorias
seg = Xu[::max(1, N // 4000)] if N > 4000 else Xu
f, pxx = welch(seg, fs=sfreq, nperseg=min(512, seg.shape[0]), axis=0)
psd = pxx.mean(axis=1)
fig, ax = plt.subplots(figsize=(11, 4))
ax.semilogy(f, psd)
ax.axvline(60, color="r", ls="--", label="60 Hz (linea)")
for lo, hi, nm in ((1, 4, "delta"), (4, 8, "theta"), (8, 13, "alpha"),
                  (13, 30, "beta"), (30, 45, "gamma")):
    ax.axvspan(lo, hi, alpha=0.10)
    ax.text((lo + hi) / 2, psd.max() * 0.4, nm, ha="center")
ax.set_xlim(0, 60); ax.set_xlabel("Hz"); ax.set_ylabel("PSD (V^2/Hz)")
ax.set_title(f"PSD media unipolar - {C} ch (bandpass 1-45 Hz en construccion)")
ax.legend(); plt.tight_layout(); plt.show()

alpha_power = psd[(f >= 8) & (f <= 13)].sum()
total_power = psd[(f >= 1) & (f <= 45)].sum()
print(f"Pico alfa presente: razon alfa/total = {alpha_power/total_power:.2%}")
print("=> Dato ya bandpaseado y con contenido fisiologico; "
      "no requiere filtraje adicional.")
```

## 4b - Validacion de las 7 referencias

* **Consistencia interna**: cada referencia almacenada debe coincidir
  con el operador analitico `build_reference_matrix` aplicado a unipolar
  (ancla): `X_k == X_unipolar @ M_k`.
* **Consistencia cruzada (49 rutas)**: el mapa analitico
  `M_{s->d} = T_d @ pinv(T_s)` cumple `X_d ~= X_s @ M_{s->d}`.
* **Propiedades fisicas**: refs sin modo comun tienen media por canal ~0;
  la columna unipolar de referencia ~0; el laplaciano es invariante a CAR.

```python
# ---- 4b - Validacion de las 7 referencias ----------------------------
from eeg_transform.references import build_reference_matrix, inter_reference_matrix

mc = data["canonical"] if "canonical" in data.order else data[data.order[0]]
split = "test"
Xu = mc.refs[split]["unipolar"]
C = mc.n_channels

# (a) cada referencia == operador analitico aplicado a unipolar (ancla)
print("=== Consistencia interna: almacenada vs operador analitico (desde unipolar) ===")
rows = []
for k in REFERENCE_KINDS:
    M = build_reference_matrix(k, C,
        unipolar_ref_index=mc.unipolar_ref_index,
        lead_field=mc.leadfield, rest_rcond=cfg.leadfield.rest_rcond,
        positions=mc.positions)
    err = Xu @ M - mc.refs[split][k]
    rows.append((k, float(np.sqrt((err ** 2).mean())), float(np.abs(err).max())))
ref_df = pd.DataFrame(rows, columns=["referencia", "rmse_uV", "max_abs_uV"]).round(5)
print(ref_df.to_string(index=False))

# (b) consistencia cruzada 49 rutas: X_d ~= X_s @ M_{s->d}
print("\n=== Consistencia cruzada (49 rutas) ===")
sd_err = np.zeros((len(REFERENCE_KINDS), len(REFERENCE_KINDS)))
for i, s in enumerate(REFERENCE_KINDS):
    for j, d in enumerate(REFERENCE_KINDS):
        M_sd = inter_reference_matrix(s, d, C,
            unipolar_ref_index=mc.unipolar_ref_index,
            lead_field=mc.leadfield, rest_rcond=cfg.leadfield.rest_rcond,
            positions=mc.positions)
        e = mc.refs[split][s] @ M_sd - mc.refs[split][d]
        sd_err[i, j] = np.sqrt((e ** 2).mean())
fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(np.log10(sd_err + 1e-12), cmap="viridis")
ax.set_xticks(range(len(REFERENCE_KINDS))); ax.set_xticklabels(REFERENCE_KINDS, rotation=45, ha="right")
ax.set_yticks(range(len(REFERENCE_KINDS))); ax.set_yticklabels(REFERENCE_KINDS)
ax.set_title("RMSE log10 (uV) ruta s->d (analitico)")
fig.colorbar(im); plt.tight_layout(); plt.show()

# (c) propiedades fisicas
print("=== Propiedades fisicas ===")
def chan_mean(X): return np.abs(X.mean(axis=1)).mean()
checks = {}
for k in ("car", "rest", "linked_mastoids", "linked_ears", "bipolar", "laplacian"):
    checks[k] = float(chan_mean(mc.refs[split][k]))
print("Media por canal (debe ~0 para refs sin modo comun):")
for k, v in checks.items():
    print(f"  {k:16s} {v:.2e} uV")
ui = mc.unipolar_ref_index
print(f"Unipolar col referencia ({ds.ch_names[ui]}): "
      f"RMS = {np.sqrt((mc.refs[split]['unipolar'][:, ui] ** 2).mean()):.2e} uV")
Mcar = build_reference_matrix("car", C, unipolar_ref_index=mc.unipolar_ref_index)
lap = mc.refs[split]["laplacian"]
inv = lap @ Mcar - lap
print(f"Laplaciano invariante a CAR: RMS diff = {np.sqrt((inv ** 2).mean()):.2e} uV")

# (d) resumen PASS/FAIL
ok_ref = ref_df["rmse_uV"].max() < 1.0
ok_cross = sd_err.max() < 1.0
ok_phys = max(checks.values()) < 1.0
print("\n=== RESUMEN ===")
print(f"Referencias internas consistentes : {'PASS' if ok_ref else 'FAIL'} "
      f"(RMSE max {ref_df['rmse_uV'].max():.4f} uV)")
print(f"Rutas cruzadas consistentes       : {'PASS' if ok_cross else 'FAIL'} "
      f"(RMSE max {sd_err.max():.3e} uV)")
print(f"Propiedades fisicas (modo comun)  : {'PASS' if ok_phys else 'FAIL'} "
      f"(max {max(checks.values()):.2e} uV)")
```

## 5 - Potencial vs Laplaciano

```python
# ---- 5 - Campo de superficie: potencial vs laplaciano ------------------
grid_px = cfg.mapping.grid_px
split, i0 = "test", 42

fig, axes = plt.subplots(2, max(1, len(data.order)),
                         figsize=(3.4 * max(1, len(data.order)), 7))
axes = np.array(axes).reshape(2, -1)
for col, label in enumerate(data.order):
    mc = data[label]
    pot = (mc.refs[split]["unipolar"][i0] @ mc.surface.T)
    lap = (mc.refs[split]["laplacian"][i0] @ mc.surface.T)
    side = int(round(np.sqrt(pot.size)))  # malla real (no depende de grid_px)
    axes[0, col].imshow(pot.reshape(side, side), cmap="RdBu_r",
                        origin="lower")
    axes[1, col].imshow(lap.reshape(side, side), cmap="RdBu_r",
                        origin="lower")
    axes[0, col].set_title(label)
axes[0, 0].set_ylabel("unipolar"); axes[1, 0].set_ylabel("laplaciano")
for ax in axes.ravel():
    ax.set_xticks([]); ax.set_yticks([])
fig.suptitle("Potencial vs CSD en la malla compartida (misma muestra)")
plt.tight_layout(); plt.show()
```

## 6 - Dimension temporal

```python
# ---- 6 - Estructura temporal (justifica la ventana dinamica) -----------
from scipy.signal import welch

sfreq = 160.0
seg = refs_test["unipolar"][i0 : i0 + 4000]

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
max_lag = 100
cols = range(0, seg.shape[1], max(1, seg.shape[1] // 6))
ac = np.stack([
    np.correlate(seg[:, c] - seg[:, c].mean(),
                 seg[: max_lag + 1, c] - seg[: max_lag + 1, c].mean()
                 )[::-1][: max_lag + 1] for c in cols])
axes[0].plot(np.arange(max_lag + 1) / sfreq * 1000,
             (ac / ac[:, :1]).T)
axes[0].set_xlabel("lag (ms)"); axes[0].set_title("Autocorrelacion (canales)")
axes[0].axvline(cfg.model.temporal_window / sfreq * 1000, color="r",
                ls="--", label=f"T_w = {cfg.model.temporal_window} muestras")
axes[0].legend(); axes[0].grid(alpha=0.3)

f, pxx = welch(seg, fs=sfreq, nperseg=256, axis=0)
axes[1].semilogy(f, pxx.mean(axis=1))
axes[1].set_xlabel("Hz"); axes[1].set_title("PSD media")
for lo, hi, nm in ((1,4,"delta"),(8,13,"alpha"),(13,30,"beta")):
    axes[1].axvspan(lo, hi, alpha=0.12)
    axes[1].text((lo+hi)/2, pxx.mean(axis=1).max()*0.3, nm, ha="center")
plt.tight_layout(); plt.show()

print(f"Cabeza dinamica: ventana {cfg.model.temporal_window} "
      f"({cfg.model.temporal_window/sfreq*1000:.0f} ms), "
      f"stride {cfg.model.temporal_stride}")
```

## Conclusiones de la exploracion

* Todas las distribuciones provienen de **electrodos reales**; las 7
  referencias son operadores lineales analiticos sobre cada casco.
* El **laplaciano** vive en otra escala espacial (alta frecuencia): mismo

  justifican la cabeza dinamica de ventana centrada: las rutas mal
  condicionadas (rest/laplaciano desde cascos escasos) son las que mas
